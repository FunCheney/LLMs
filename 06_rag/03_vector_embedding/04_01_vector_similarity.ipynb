```python
import numpy as np
from numpy import dot
from numpy import linalg

# 余弦相似度
def cos_similarity(a, b):
    return dot(a, b) / (linalg.norm(a) * linalg.norm(b))

# l2 范数
def l2_similarity(a, b):
    x = np.asarray(a) - np.asarray(b)
    return linalg.norm(x)
```

```python
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()

client = OpenAI(api_key=os.getenv("BASE_EMBEDDINGS_API_KEY"), base_url=os.getenv("BASE_EMBEDDINGS_URL"))

def get_embeddings(input: str, model='text'):
    data = client.embeddings.create(input=input, model=model)
    return [x.embedding for x in data]

query = ""
documents = []

# 把用户问题转成向量
query_vector = get_embeddings(query)
# 把文档转成向量
doc_vectors = get_embeddings(documents)
# 用户向量 和 文档向量相似度计算
print(f'余弦相似度计算。值越大越相似')
for doc_vector in doc_vectors:
    print(f'用户问题 与 文档向量相似度 {cos_similarity(query_vector, doc_vector)}')

print(f'范数相似度。值越小越相似')
for doc_vector in doc_vectors:
    print(f'用户问题 与 文档向量相似度 {l2_similarity(query_vector, doc_vector)}')
```
