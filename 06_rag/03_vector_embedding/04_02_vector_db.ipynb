```python
import chromadb

# 创建连接
client = chromadb.Client()
# 持久化存储
chromadb.PersistentClient(path="./chroma_db")
# 获取集合
collection = client.get_or_create_collection('test')

# 增加
collection.add(
    documents=[""], # 文本内容列表，每个元素是一段文本
    embeddings=[[1,2,3],[4,5,6],[7,8,9]], # 嵌入向量列表，每个元素是一个与 documents 对应的向量表示
    ids=["1","2","3"] # 自定义 ID 列表，用于唯一标识每条记录
)
# 存储
q = collection.query(
    ids=["1"],
    include=['documents', 'embeddings'],
)
print(q)

q1 = collection.query(
    where_document={'$contains': '你好'}
)
print(q1)
# 删除
collection.delete(
    ids=['2']
)
print(collection.get())

# 修改
collection.update(
    documents=[""],
    embeddings=[[1,20,3],[4,50,6],[7,8,90]],
    ids=["1","2","3"],
)
print(collection.get(include=["embeddings"]))
```
