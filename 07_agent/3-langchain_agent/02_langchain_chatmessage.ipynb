```python
import os
from dotenv import load_dotenv
load_dotenv()
model_name = os.getenv("LOCAL_MODE")
base_url = os.getenv("LOCAL_BASE_URL")
```

```python
from langchain_ollama import OllamaLLM
llm = OllamaLLM(
    model = model_name,
    base_url=base_url,
    temperature=0.2
)
```

```python
from langchain_core.messages import HumanMessage
human_msg = HumanMessage(
    content="你好！我想学习 Python 编程，有什么建议吗？"
)

print(human_msg.type)
print(human_msg.content)
```

```python
resp = llm.invoke([
    human_msg
])
print(resp)
```

```python
from langchain_core.messages import SystemMessage

resp = llm.invoke([
    SystemMessage(content="你是一个天气智能助手"),
    HumanMessage(content="我想查询北京今天的天气")
])

print(resp)
```

```python
from langchain_core.messages import FunctionMessage
import json, datetime
# 模拟获取当前时间
current_time = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
function_msg = FunctionMessage(
    content=json.dumps({"result": current_time, "timezone": "Asia/Shanghai"}, ensure_ascii=False),
    name="get_current_time"
)
print(function_msg.name)
print(function_msg.content)
print(function_msg.type)
```
