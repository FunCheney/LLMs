### API 调用
1. zeros() 返回给定形状和类型的新数组，用 0 填充
2. ones() 返回给定形状和类型的数组，用 1 填充
3. empty() 返回给定形状和类型的为初始化的新数组

```python
import numpy as np
arr = np.zeros((2,3))
print(arr)
```

```python
arr = np.ones((2,3))
arr
```

```python
arr1 = np.empty((2,3))
arr1
```

1. zeros_like()
2. ones_like()
3. empty_like()

```python
arr3 = np.zeros_like(arr1)
arr3
```

```python
arr4 = np.ones_like(arr1)
arr4
```

```python
arr5 = np.empty_like(arr)
arr5
```

1. full()
2. full_like()

```python
arr1 = np.full((2,3), 5)
print(arr1)
arr2 = np.full_like(arr1, 3)
print(arr2)
```

```python
arr = np.arange(0, 10, 2)
arr
```

```python
arr = np.arange(0, 2, 0.3)
print(arr)
```

```python
from numpy import pi
arr = np.linspace(0, 2, 9) # 0 - 2 之间生成 9 个数
print(arr)

x = np.linspace(0, 2 * pi, 100)
print(np.sin(x))
```

```python
arr = np.arange(12)
print(arr)
arr1 = arr.reshape(3, 4)
print(arr1)
arr2 = np.arange(24).reshape(2, 3, 4)
print(arr2)
```
