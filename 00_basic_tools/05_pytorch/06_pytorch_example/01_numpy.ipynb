```python
import numpy as np
import math

from matplotlib import pyplot as plt

# 在[-π, π] 区间 生成 2000 个均匀分布的点
x = np.linspace(-math.pi, math.pi, 2000)
# 计算对应 x 真的正弦值作为目标输出
y = np.sin(x)
# 绘图
plt.plot(x, y)
```

```python
# 初始化权重,随机初始化多项式系数
a = np.random.randn()
b = np.random.randn()
c = np.random.randn()
d = np.random.randn()

leaning_rate = 1e-6
```

```python
for t in range(2000):
    # Forward pass: compute predicted y
    # y = a + b x + c x^2 + d x^3
    y_pred = a + b * x + c * x **2 + d * x ** 3

    # 计算并打印损失, 使用均方误差作为损失函数
    loss = np.square(y_pred - y).sum()
    if t % 100 == 99:
        print(t, loss)

    # 反向传播：计算各参数的梯度
    grad_y_pred = 2.0 * (y_pred - y) # 损失对预测值的梯度
    grad_a = grad_y_pred.sum() # 损失对 a 的梯度
    grad_b = (grad_y_pred * x).sum() # 损失对 b 的梯度
    grad_c = (grad_y_pred * x ** 2).sum() # 损失对 c 的梯度
    grad_d = (grad_y_pred * x ** 3).sum() # 损失对 d 的梯度

    # 更新参数（梯度下降更新规则）
    a -= leaning_rate * grad_a
    b -= leaning_rate * grad_b
    c -= leaning_rate * grad_c
    d -= leaning_rate * grad_d

print(f'Result: y = {a} + {b} x + {c} x^2 + {d} x^3')
```
