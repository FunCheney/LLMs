一个三阶多项式，通过最小化欧氏距离平方来训练，用于预测从−π 到 π 范围内的 y=sin (x)。该实现使用 PyTorch 张量来手动计算前向传播、损失和反向传播。

```python
import torch
import math

dtype = torch.float
device = torch.device("cpu")

x = torch.linspace(-math.pi, math.pi, 2000, device=device, dtype=dtype)
y = torch.sin(x)

a = torch.randn((), device=device, dtype=dtype)
b = torch.randn((), device=device, dtype=dtype)
c = torch.randn((), device=device, dtype=dtype)
d = torch.randn((), device=device, dtype=dtype)

learning_rate = 1e-6

for t in range(2000):

    y_pred = a + b * x + c * x ** 2 + d * x ** 3

    # y_pred 是模型的预测结果（由多项式 a + bx + cx² + dx³ 计算得出）
    # y 是真实值（这里是 sin(x) 的结果）
    # 两者相减得到一个误差张量，形状与 x 相同（2000 个元素），每个元素代表对应位置的预测误差（预测值 - 真实值）
    # sum() 将张量中所有元素（2000 个平方误差）相加，得到一个标量张量（只有一个数值）
    # 这个值代表所有样本的平方误差总和，反映了模型在所有数据上的整体误差大小
    # PyTorch 的张量（即使是标量张量）本质上是一个带计算图信息的对象
    # .item() 方法可以将标量张量转换为 Python 原生的数值类型（如 float），方便后续的打印、存储等操作
    # 如果直接使用张量进行打印或运算，可能会附带额外的张量元信息，而 .item() 能得到纯净的数值
    loss = (y_pred - y).pow(2).sum().item()

    if t % 100 == 0:
        print(t, loss)


    grade_y_pred = 2 * (y_pred - y)

    grad_a = grade_y_pred.sum()
    grad_b = (grade_y_pred * x).sum()
    grad_c = (grade_y_pred * x ** 2).sum()
    grad_d = (grade_y_pred * x ** 3).sum()


    a -= learning_rate * grad_a
    b -= learning_rate * grad_b
    c -= learning_rate * grad_c
    d -= learning_rate * grad_d

print(f'Result: y = {a.item()} + {b.item()} x + {c.item()} x^2 + {d.item()} x^3')
```
