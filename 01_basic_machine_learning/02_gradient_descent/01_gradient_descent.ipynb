### 梯度下降
#### 目的
梯度下降法是一个算法，不是像多元线性回归那样是一个具体做回归任务的算法，而是一个非常通用的优化算法来帮助一些机器学习算法来求解出最优解的。
优化算法的目的：期望以最快的速度把模型的参数 θ 求解出来

#### 原因
1. 之前利用 θ 的解析解公式求解出出来的解，我们直接说是最优也的原因是因为 MSE 一个凸函数。如果是非凸函数，设置梯度为 0 会得到多个极值。
2. 当数据的特征纬度变多之后，解析解的公式 X^T * X 是 n 维乘以 n 维的。复杂度是 O(n^3)

#### 公式
通常表示为：
θ_{t+1} = θ_t - η ∇J(θ_t)

其中：
1. θ 是参数向量
2. η 是学习率

   2.1 设置比较大

   2.2 设置比较小

3. ∇J(θ) 是目标函数 J 关于 θ 的梯度
